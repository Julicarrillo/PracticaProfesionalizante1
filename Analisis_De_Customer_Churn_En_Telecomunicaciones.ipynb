# Desarrollo del Análisis de Customer Churn en Telecomunicaciones

## Abstract



El presente trabajo analiza el problema de abandono de clientes (*customer churn*) en una empresa de telecomunicaciones a partir del dataset público **IBM Telco Customer Churn**, desarrollado por IBM como parte de su plataforma de analítica Cognos Analytics. El conjunto de datos registra información de 7.043 clientes de una compañía ficticia que proveyó servicios de telefonía e internet en el estado de California, Estados Unidos, durante el tercer trimestre de un año fiscal. Cada registro representa un cliente único y contiene 21 atributos que describen su perfil demográfico, los servicios contratados, las condiciones de su cuenta y si efectivamente abandonó el servicio durante el período observado.

El objetivo central del análisis es identificar qué características de los clientes se asocian con una mayor probabilidad de abandono, con el propósito de sentar las bases para un modelo predictivo que permita anticipar la fuga antes de que ocurra. Este tipo de problema tiene aplicación directa en cualquier empresa de servicios con cartera de clientes activa, donde retener un cliente existente representa un costo significativamente menor que adquirir uno nuevo.

Entre las variables disponibles, se consideran de especial interés el tipo de contrato (`Contract`), la antigüedad del cliente (`tenure`), el cargo mensual (`MonthlyCharges`) y los servicios adicionales contratados, dado que se hipotetiza que estas dimensiones concentran la mayor parte del poder explicativo sobre el comportamiento de abandono.

El análisis exploratorio de datos incluirá el estudio de distribuciones univariadas, el análisis de variables categóricas en relación al target, la detección de valores nulos con lógica de negocio —en particular el campo `TotalCharges`, que presenta entradas vacías para clientes con antigüedad cero— y la exploración de correlaciones entre variables numéricas. Previo al modelado, se realizará un proceso de limpieza y preprocesamiento que contempla la conversión de tipos de datos, la codificación de variables categóricas según su naturaleza ordinal o nominal, y la normalización de variables numéricas.

Como proyección hacia la siguiente fase del proyecto, se plantea la construcción de un modelo de clasificación supervisada para predecir la variable `Churn`.

```python
#@title 2 - Distribución de abandono según antigüedad

#@markdown 2.1 Cual es la distribucion de abandono de los clientes?


df_telco =df_telco.copy()

df_churn = df_telco[df_telco['Churn'] == 'Yes']

media = df_churn['tenure'].mean()
mediana = df_churn['tenure'].median()

fig = px.histogram(
    df_churn,
    x='tenure',
    nbins=30,
    histnorm='percent',
    template='plotly_dark',
    color_discrete_sequence=['#FFB347']
)

fig.update_layout(
    yaxis_title='Porcentaje de abandono (%)',
    xaxis_title='Antigüedad (en meses)',
    bargap=0.08,
    title={
        'text': '📊Distribución de abandono según antigüedad',
        'x': 0.5,
        'xanchor': 'center',
        'font': {
            'size': 28,
            'color': '#FFF9C4'
        }
    }
)

fig.update_yaxes(
    ticksuffix='%'
)

fig.update_xaxes(
    tickmode='linear',
    dtick=6
)

fig.update_traces(
    xbins=dict(
        start=1,
        end=72,
        size=2  # intervalos de 2 meses
    ),
    marker_line_color='white',
    marker_line_width=0.8,
    hovertemplate=
    'Mes %{x}<br>' +
    'Porcentaje: %{y:.2f}%' +
    '<extra></extra>'
)

#Media
fig.add_vline(
    x=media,
    line_dash='dash',
    line_color='#d73027',
    line_width=2,
    annotation_text=f"Media: {media:.1f} meses",
    annotation_position="top",
    annotation_font=dict(color='#d73027', size=14)
)

#Mediana
fig.add_vline(
    x=mediana,
    line_dash='dot',
    line_color='#d73027',
    line_width=2,
    annotation_text=f"Mediana: {mediana:.1f}",
    annotation_position="top",
    annotation_font=dict(color='#d73027')
)
fig_tenure = fig
fig.show()

```

```python
#@title  3 - Porcentaje de clientes que abandonaron según tipo de contrat

#@markdown 3.1 Que tipo de contrato tienen los clientes que abandonan?

df =df_telco.copy()

# Total de clientes por tipo de contrato
total = df.groupby('Contract').size().reset_index(name='Total')

# Clientes que abandonaron por tipo de contrato
churn = (
    df[df['Churn'] == 'Yes']
    .groupby('Contract')
    .size()
    .reset_index(name='Clientes que abandonaron')
)

# Unimos ambas tablas
df_contract = total.merge(churn, on='Contract', how='left')

# Si algún contrato no tiene churn, lo rellenamos con 0
df_contract['Clientes que abandonaron'] = df_contract['Clientes que abandonaron'].fillna(0)

# Calculamos porcentaje de churn
df_contract['Porcentaje churn'] = (
    df_contract['Clientes que abandonaron'] / df_contract['Total']
) * 100

# Ordenamos de mayor a menor porcentaje
df_contract = df_contract.sort_values(by='Porcentaje churn', ascending=False)

# Creamos texto para mostrar al pasar el cursor
df_contract['Detalle'] = (
    df_contract['Clientes que abandonaron'].astype(int).astype(str)
    + ' de '
    + df_contract['Total'].astype(int).astype(str)
    + ' clientes abandonaron ('
    + df_contract['Porcentaje churn'].round(2).astype(str)
    + '%)'
)

df_contract['Contract'] = df_contract['Contract'].replace({
    'Month-to-month': 'Mes a mes',
    'One year': 'Anual',
    'Two year': 'Cada 2 años'
})

fig = px.bar(
    df_contract,
    x='Contract',
    y='Porcentaje churn',
    text='Porcentaje churn',
    color='Porcentaje churn',
    color_continuous_scale='RdYlGn_r',
    hover_name='Contract',
    hover_data={
        'Porcentaje churn': ':.2f',
        'Clientes que abandonaron': True,
        'Total': True,
        'Detalle': True
    },
    title='📉 Porcentaje de clientes que abandonaron según tipo de contrato'
)

fig.update_traces(
    texttemplate='%{text:.2f}%',
    textposition='outside',
    hovertemplate='<b>%{x}</b><br>' + '%{customdata[2]}' + '<extra></extra>'
)

fig.update_layout(
    template='plotly_dark',
    title={
        'x': 0.5,
        'xanchor': 'center',
        'font': {
            'size': 24,
            'color': '#FFF9C4'
        }
    },
    xaxis_title=' ',
    yaxis_title='Porcentaje de churn (%)',
    showlegend=False,
    coloraxis_showscale=False
)

fig.update_yaxes(
    ticksuffix='%',
    range=[0, df_contract['Porcentaje churn'].max() + 10]
)
fig_contract = fig
fig.show()

```

Este gráfico complementa el análisis anterior, ya que no muestra solamente la distribución de los clientes que abandonaron, sino la tasa de churn dentro de cada rango de antigüedad.

Se observa que los clientes con menor antigüedad presentan una mayor proporción de abandono, especialmente durante los primeros meses de permanencia. A medida que aumenta la antigüedad del cliente, la tasa de churn disminuye, lo que refuerza la hipótesis de que los primeros meses de relación con la empresa son el período más crítico para implementar estrategias de retención.

```python
#@title 4 - Porcentaje de clientes que abandonaron según cantidad de servicios

#@markdown  4.1 - Que cantidad de servicios tienen aquellos clientes que abandonaron?

df_servicios = df_telco.copy()

#Tomamos las columnas de los diferentes servicios que se pueden contratar, menos el servicio de internet
servicios = [
    'PhoneService',
    'MultipleLines',
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies'
]

#Como cada servicio guarda yes, no o no internet service, la pasamos a variable binaria (1 = tiene servicio, 0 = no)
for col in servicios:
    df_servicios[col] = df_servicios[col].apply(lambda x: 1 if x == 'Yes' else 0)

#InternetService se trata aparte xq sus variables son fibra optica, DSL o No, entonces fibra optica y DSL deben ser Yes
df_servicios['InternetService_bin'] = df_servicios['InternetService'].apply(lambda x: 0 if x == 'No' else 1)

#Se filtran los clientes sin internet
df_servicios = df_servicios[df_servicios['InternetService'] != 'No']

#Defino la variable: cantidad total de servicios
df_servicios['TotalServicios'] = df_servicios[servicios].sum(axis=1) + df_servicios['InternetService_bin']

#Agrupa por cantidad de servicios y cuenta cuantos clientes hay
total = df_servicios.groupby('TotalServicios').size().reset_index(name='Total')

#Filtra los clientes que abandonaron y cuantos hay por grupo
churn = df_servicios[df_servicios['Churn'] == 'Yes'].groupby('TotalServicios').size().reset_index(name='Cantidad')

#Une ambos datos - Usamos how='left' para conservar todas las categorías de cantidad de servicios
df_churn = total.merge(churn, on='TotalServicios', how='left')

#Si algún servicio no tiene churn, lo rellenamos con 0
df_churn['Cantidad'] = df_churn['Cantidad'].fillna(0)

#Calcula el porcentaje de abandono
df_churn['Proporcion'] = (df_churn['Cantidad'] / df_churn['Total']) * 100


fig = px.bar(
    df_churn,
    x='TotalServicios',
    y='Proporcion',
    text='Proporcion',
    title='📊Porcentaje de clientes que abandonaron según cantidad de servicios',
    template='plotly_dark'
)

fig.update_traces(
    marker=dict(
        color=df_churn['TotalServicios'],
        colorscale='RdYlGn'
    ),
    texttemplate='%{text:.1f}%',
    textposition='outside',
    customdata=df_churn[['Cantidad', 'Total']],
    hovertemplate=
    '<b>Servicios:</b> %{x}<br>' +
    '<b>Abandono:</b> %{y:.1f}%<br>' +
    '<b>%{customdata[0]} de %{customdata[1]} clientes abandonaron</b><extra></extra>'
)

fig.update_layout(
    title=dict(
        font=dict(size=24, color='#FFF3B0')
    ),
    xaxis_title='Cantidad de servicios contratados',
    yaxis_title='Proporción (%)'
)

fig_servicios = fig
fig.show()
```

Para este análisis se consideraron únicamente los clientes que poseen servicio de internet, ya que varias de las variables analizadas corresponden a servicios asociados a internet, como seguridad online, backup online, soporte técnico, streaming TV y streaming movies.

Además, al unir la tabla con el total de clientes por cantidad de servicios y la tabla de clientes que abandonaron, se utilizó un merge con how='left'. Esto permite conservar todas las categorías de cantidad de servicios, incluso aquellas que eventualmente no presenten casos de churn.

```python
#@title 7 - Porcentaje de abandono según servicios y cargos mensuales

#@markdown 7.1 Como es el % de abandono por la cantidad de servicios contratados y cargos mensuales?

df = df_telco.copy()

#Tomamos las columnas de los diferentes servicios que se pueden contratar, menos el servicio de internet
servicios = [
    'PhoneService',
    'MultipleLines',
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies'
]

#Como cada servicio guarda yes, no o no internet service, la pasamos a variable binaria (1 = tiene servicio, 0 = no)
for col in servicios:
    df[col] = df[col].apply(lambda x: 1 if x == 'Yes' else 0)

#InternetService se trata aparte xq sus variables son fibra optica, DSL o No, entonces fibra optica y DSL deben ser Yes
df['InternetService_bin'] = df['InternetService'].apply(lambda x: 0 if x == 'No' else 1)

#Se filtran los clientes sin internet
df = df[df['InternetService'] != 'No']

#Defino la variable: cantidad total de servicios
df['TotalServicios'] = df[servicios].sum(axis=1) + df['InternetService_bin']
# Rangos de cargos mensuales
df['RangoCargos'] = pd.cut(
    df['MonthlyCharges'],
    bins=6
)

# Guarda el orden correcto de los rangos ANTES de convertirlos a texto
orden_rangos = sorted(df['RangoCargos'].dropna().unique(), reverse=True)

# Total de clientes por grupo
total_2 = (
    df.groupby(['TotalServicios', 'RangoCargos'], observed=False)
    .size()
    .reset_index(name='Total')
)

# Clientes que abandonaron por grupo
churn_2 = (
    df[df['Churn'] == 'Yes']
    .groupby(['TotalServicios', 'RangoCargos'], observed=False)
    .size()
    .reset_index(name='Cantidad')
)

# Unir
df_heat = total_2.merge(
    churn_2,
    on=['TotalServicios', 'RangoCargos'],
    how='left'
)

# Reemplazar NaN por 0
df_heat['Cantidad'] = df_heat['Cantidad'].fillna(0)

# Calcular porcentaje
df_heat['Proporcion'] = (df_heat['Cantidad'] / df_heat['Total']) * 100

# Mantener los rangos ordenados correctamente
df_heat['RangoCargos'] = pd.Categorical(
    df_heat['RangoCargos'],
    categories=orden_rangos,
    ordered=True
)

# Crear tabla pivot para heatmap
pivot = df_heat.pivot(
    index='RangoCargos',
    columns='TotalServicios',
    values='Proporcion'
)

pivot.index = pivot.index.astype(str)

fig = px.imshow(
    pivot,
    color_continuous_scale='RdYlGn_r',
    text_auto='.2f',
    aspect='auto',
    template='plotly_dark'
)

fig.update_layout(
    title={
        'text': '🧮Porcentaje de abandono según servicios y cargos mensuales',
        'x': 0.5,
        'xanchor': 'center',
        'font': {
            'size': 26,
            'color': '#FFF3B0'
        }
    },
    xaxis_title='Cantidad de servicios contratados',
    yaxis_title='Rango de cargos mensuales (en USD)',
    coloraxis_colorbar_title='Abandono (%)'
)

fig.update_traces(
    hovertemplate=
    '<b>Servicios:</b> %{x}<br>' +
    '<b>Rango:</b> %{y}<br>' +
    '<b>Abandono:</b> %{z:.2f}%<extra></extra>'
)

fig_heatmap = fig
fig.show()
```

# Selección del Modelo de Machine Learning



## Objetivo

A partir de los resultados obtenidos en el análisis exploratorio, se plantea
como siguiente etapa el desarrollo de un modelo de Machine Learning capaz de
predecir si un cliente abandonará el servicio (`Churn = Yes`) o permanecerá
activo (`Churn = No`).

La variable objetivo es **Churn**, por lo que se trata de un problema de
**clasificación binaria supervisada** — el modelo aprende a partir de ejemplos
históricos etiquetados para luego predecir casos nuevos.

---

## Consideración sobre el desbalance de clases

El dataset presenta un desbalance moderado: **73.4% No churn / 26.6% Churn**.
Esto tiene una implicancia directa en la evaluación del modelo: la métrica
**Accuracy** no es adecuada en este contexto, ya que un modelo que prediga
siempre "No churn" alcanzaría un 73% de exactitud sin identificar ningún
cliente en riesgo.

Por este motivo, las métricas de evaluación propuestas son:

- **AUC-ROC** — mide la capacidad del modelo de separar ambas clases
independientemente de su distribución.
- **Recall (clase positiva)** — prioriza la detección de clientes que
realmente van a abandonar. El costo de no detectar un churn real
es mayor que el de una falsa alarma.
- **F1-Score** — balance entre Precision y Recall.

## Preparación de los datos para Machine Learning

Los algoritmos de aprendizaje automático requieren que todas las variables
sean numéricas y estén en escalas comparables. Por este motivo se aplicaron
tres técnicas de transformación sobre una copia del dataset limpio,
sin modificar los datos utilizados en el EDA.

**Label Encoding:** para variables binarias con dos categorías (Yes/No,
Male/Female). Se asigna 0 a un valor y 1 al otro de forma directa.

**One Hot Encoding:** para variables categóricas nominales con más de dos
categorías (Contract, InternetService, PaymentMethod, etc.). Se crea una
columna binaria por cada categoría. Se usa `drop_first=True` para evitar
multicolinealidad — una de las categorías queda implícita en las demás.

**Normalización Z-score (StandardScaler):** para variables numéricas con
distintas escalas (tenure en meses, MonthlyCharges en USD). La normalización
lleva cada variable a media 0 y desvío estándar 1, evitando que variables
de mayor magnitud dominen el modelo.

```python
df_ml = df_telco.copy()

# Eliminamos customerID porque es solo un identificador, no sirve para entrenar
df_ml = df_ml.drop(columns=['customerID'])

# LABEL ENCODING
binarias = [
    'gender', 'SeniorCitizen', 'Partner', 'Dependents',
    'PhoneService', 'PaperlessBilling', 'Churn'
]

for col in binarias:
    df_ml[col] = df_ml[col].map({
        'Yes': 1, 'No': 0,
        'Female': 1, 'Male': 0
    })

# ONE HOT ENCODING
columnas_categoricas = [
    'MultipleLines', 'InternetService', 'OnlineSecurity',
    'OnlineBackup', 'DeviceProtection', 'TechSupport',
    'StreamingTV', 'StreamingMovies', 'Contract', 'PaymentMethod'
]

df_ml = pd.get_dummies(df_ml, columns=columnas_categoricas, drop_first=True)

# Convertimos True/False a 0/1
df_ml = df_ml.astype(int, errors='ignore')

# NORMALIZACIÓN Z-SCORE
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
columnas_numericas = ['tenure', 'MonthlyCharges', 'TotalCharges']
df_ml[columnas_numericas] = scaler.fit_transform(df_ml[columnas_numericas])

# Verificación: no deberían quedar columnas de texto
print(df_ml.select_dtypes(include='object').columns)

df_ml.head()
```
